4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

*  Mostra la resposta JSON.
*  Mostra el codi d'estat.

```python
import requests
import json

url = 'https://jsonplaceholder.typicode.com/posts'
payload = {
    'title': 'publicació fictícia',
    'body': 'crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId',
    'userId': 55
}
headers = {
    'Content-type': 'application/json; charset=UTF-8'
}

response = requests.post(url, data=json.dumps(payload), headers=headers)

print(f"Codigo estado: {response.status_code}")
print("Respuesta JSON:")
print(response.json())
```

5. Fes una petició PATCH per modificar parcialment una publicació existent.

*  Mostra la resposta JSON.
*  Mostra el codi d'estat.

```python
import requests
import json

url_patch = 'https://jsonplaceholder.typicode.com/posts/101'

# Payload para modificar parcialmente (solo cambiamos el título)
payload_patch = {
    'body': 'modificar parcialment una publicació existent.'
}

headers = {
    'Content-type': 'application/json; charset=UTF-8'
}

response_patch = requests.patch(url_patch, data=json.dumps(payload_patch), headers=headers)

print(f"Codi d'estat: {response_patch.status_code}")
print("Resposta JSON:")
print(response_patch.json())
```

6. Fes una petició DELETE sobre una publicació.

*  Mostra la resposta JSON.
*  Mostra el codi d'estat.

```python
import requests

url_delete = 'https://jsonplaceholder.typicode.com/posts/101'

# Realizar la petición DELETE
response_delete = requests.delete(url_delete)

print(f"Codi d'estat: {response_delete.status_code}")
print("Resposta JSON:")
print(response_delete.json())
```

Weatherstack provides several endpoints to access different types of weather data:

*  Current Weather: Retrieve current weather conditions.
*  Historical Weather: Access past weather data for a specific date.
*  Historical Time-Series: Retrieve weather data over a range of dates.
*  Weather Forecast: Get forecasts for up to 14 days. http://api.weatherstack.com/forecast

https://weatherstack.com/parameters

*  Header Parameters
These parameters are frequently connected to permission and are found in the request header.

*  Query Parameters
A question mark separates the query parameters from the hierarchy parameters.

*  Request Body Parameters
Request body parameters are used to send and receive data using the RESTful API and are contained in the request body.

*  Parameters for the Template/Path
Route parameters are placed within the route of an endpoint just before the query string and offer developers an effective approach to parameterize assets. They are set off within curly brackets.

*  Matrix Parameters
Matrix parameters sit between the resource path and query parameters, divided by a semicolon from the hierarchical parameters.

*  Plain Parameters
Plain parameters are request parameters that are readily available in ReadyAPI but are removed when the request is sent.

DEVOLVER DATAFRAME CON TODAS LAS COLUMNAS

Datos anidados (Estructuras JSON complejas)

Si el JSON contiene objetos o diccionarios dentro de otros, pd.DataFrame() creará columnas que contienen diccionarios completos. Para "aplanar" esa estructura en columnas individuales automáticas, se utiliza pd.json_normalize()

```python
import pandas as pd
import requests

url = "http://api.weatherstack.com/current"

# parámetros de consulta
parametros = {
    'access_key': 'dfc36df7240b4559e918d844001178a9',
    'query': 'Paris'
}

# petición HTTP GET
resp = requests.get(url, params=parametros)

# 1. Convertir la respuesta a JSON (dict/lista de Python)
datos = resp.json()

# 2. Crear el DataFrame aplanando los diccionarios anidados
df = pd.json_normalize(datos["current"])
display(df.head())
```
