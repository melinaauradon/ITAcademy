> Corregir possibles problemes de codificació i accentuació en els noms de les columnes i en els valors de text, si escau.

Hice algunos intentos y me encontre con varios problemas:

1. Primero empece usando el metodo str.encode, pero el dataframe tambien contiene valores numericos y booleanos. Debemos ignorarlos.
2. Trate de re-codificar a latin-1, pero algunos caracteres no estaban contemplados. Usare cp1252.
3. Intente arreglar todo el dataframe de golpe, pero al haber celda corectas e incorectas mezcladas, generaba fallos. Procederemos elemento por elemento y si una celda es correcta la dejamos intacta.

Por otro lado, al habar detectado unicamente un nombre de columna con error de codificacion, la modificaremos directamente.

```python
df = df.rename(columns={"CÃ\xa0rrec": "Càrrec"})
```

```python
df = df.rename(columns={"Unnamed: 0": "ID"})
```

```python
def corregir_mojibake(valor):
    # Si el valor no es string (NaN, numeros), se descarta.
    if not isinstance(valor, str):
        return valor

    try:
        # Se recodifica en CP1252 y decodifica en UTF-8
        return valor.encode('cp1252').decode('utf-8')
    except (UnicodeEncodeError, UnicodeDecodeError):
        # Si el valor está bien escrito dará error al intentar re-codificar; en ese caso lo dejamos tal cual
        return valor

# Aplicamos la corrección unicamente a las columnas de tipo texto (object)
columnas_texto = df.select_dtypes(include=['object']).columns

for col in columnas_texto:
    try:
        # Aplicamos el cambio celda por celda con .map()
        df[col] = df[col].map(corregir_mojibake)

        print(f"'{col}' corregida.")
    except Exception as e:
        print(f" Error '{col}': {e}")
```

**Salari mensual**

Revisamos primero que contiene la columna Salari mensual.
Vemos que tiene palabras, simbolos..

Crearemos une nueva columna, empezamos limpiando los espacios, comas, puntos, el simbolo € y la palabra 'euros'.

```python
df['Salari mensual'].isna().sum()
```

```python
df['Salari mensual'].sample(20)
```

```python
# 1. Creamos la nueva columna como una copia de la original
df['Salari mensual_euros'] = df['Salari mensual'].astype(str)

# 2. Definimos el patrón de limpieza (espacios, €, comas, puntos y la palabra 'euros')
patron_limpieza = r'[\s€,.]|euros'

# 3. Aplicamos la limpieza en la nueva columna
df['Salari mensual_euros'] = df['Salari mensual_euros'].str.replace(patron_limpieza, '', regex=True, case=False)
```

Despues de esta primera limpieza, comprobamos que valores nos quedan por contemplar

```python
# Filtramos las celdas que no son números tras la limpieza
filas_con_texto = df[~df['Salari mensual_euros'].str.isdigit()][['Salari mensual_euros']]

# Mostramos los valores únicos para ver qué palabras quedaron
print(list(filas_con_texto['Salari mensual_euros'].unique()))
```

Cambiamos las palabras por numeros o por NaN segun el caso y aplicamos la conversion a tipo numerico

```python
# 4. Diccionario de correspondencia adaptado (sin espacios ni símbolos)
mapping = {
    'milvuit-cents': '1800',
    'dosmil': '2000',
    'tresmil': '3000',
    'molts': np.nan,
    'n/d': np.nan,
    'nan': np.nan
}

# 5. Reemplazamos el texto por los números correspondientes en la nueva columna
df['Salari mensual_euros'] = df['Salari mensual_euros'].replace(mapping)

# 6. Convertimos el resultado final a tipo numérico de forma definitiva
df['Salari mensual_euros'] = pd.to_numeric(df['Salari mensual_euros'], errors='coerce')
```

**franjas de edad:** Crearemos una nueva columna para categorizar la edad de los empleados según las franjas especificadas en politica_salarial.

Se debe crear primero la fecha de nacimiento a partir de las 3 columnas dia, mes y any de naixement.

Despues calcumos la edad que es la resta entre el dia de hoy y la fecha de nacimiento.

Por ultimo se crea la columna Franja_edat y se asigna una de las 4 franjas.

```python
# Creamos una fecha de nacimiento temporalmente
fecha_nacimiento = pd.to_datetime(
    {
        "year": df2["Any de Naixement"],
        "month": df2["Mes de Naixement"],
        "day": df2["Dia de Naixement"]
    },
    errors="coerce"
)

# Calculamos la edad: contempla años bisiestiles y que devuelva un numero entero inferior (si 20,6, guardamos 20 años)
hoy = pd.Timestamp.today()

df2["Edat"] = np.floor(
    (hoy - fecha_nacimiento).dt.days / 365.25 # "365.25"contempla años bisiestiles y np.floor devuelve la parte entera del decimal (si 20,6, guardamos 20 años)
).astype("Int64")

# Definimos las franjas de edad
condiciones = [
    df2["Edat"].between(18, 29).fillna(False),
    df2["Edat"].between(30, 44).fillna(False),
    df2["Edat"].between(45, 59).fillna(False),
    (df2["Edat"] >= 60).fillna(False)
]

franjas = ["18-29", "30-44", "45-59", "60+"]

df2["Franja edat"] = np.select(
    condiciones,
    franjas,
    default=pd.NA
)
```

```python
base_increment_by_group = {
    'Grup A': 5.0,
    'Grup B': 3.5,
    'Grup C': 2.0,
    'Grup D': 8.0
}

age_adjustment = {
    '18-29': 0.0,
    '30-44': 1.0,
    '45-59': 2.0,
    '60+': 1.5
}

cost_of_living_adjustment = {
    'Low': 0.0,
    'Medium': 1.0,
    'High': 2.0
}
```

-'inc_grup' para Grup professional

-'inc_fills' para Situació familiar

-'inc_edat' para Franja d'edat

-'inc_cost_vida' para Cost de vida de la ciutat

```python
df2['inc grup'] = df2['Grup Professional'].map(base_increment_by_group)


df2['inc fills'] = np.where(
    df2['Nombre_fills'].isna(), # condición: si falta el numero de hijos
    np.nan, # entonces el incremento es NaN
    np.where(df2['Nombre_fills'] > 0, 1.5, 0.0) # si existe, se calcula el incremento
)

df2['inc edat'] = df2['Franja edat'].map(age_adjustment)

df2['inc cost vida'] = df2['Ajust cost vida'].map(cost_of_living_adjustment)
```

Calculamos el nuevo salario

La política salarial proporciona los porcentajes asociados a las categorías de transporte público (low, medium y high), pero no define como asignar cada ciudad a una categoría concreta. Dado que no existe ningun dato oficial (como era el caso para el indice de coste de vida, aunque "semi-official") ni tampoco una regla objetiva que determine los valores de las categorías, preferimos no incluir este ajuste en el cálculo salarial para evitar introducir supuestos arbitrarios que podrían alterar los resultados finales.

No obstante calculamos de forma exploratoria un posible ajuste comparando dos metodos, terciles vs rupturas naturales.

```python
# Ajuste aplicando Terciles (Divide el total en tres partes iguales)
transport_adjustment = {
    'low': 0.5,
    'medium': 0.0,
    'high': -0.5
}

# Clasificación en 3 grupos
ciutats['Cat_transport_terciles'] = pd.qcut(
    ciutats['Transport_public_score'],
    q=3,
    labels=['low', 'medium', 'high']
)
ciutats['inc_transport_terciles'] = (ciutats['Cat_transport_terciles'].map(transport_adjustment))
```

Comparativa entre terciles y Jenks:

El método de terciles divide las ciudades en tres grupos aproximadamente iguales (no son 3 grupos exactamente iguales debido a que hay empates entre ciudades) según su posición dentro de la distribución. En cambio, el método de Jenks busca identificar agrupaciones naturales, estableciendo los puntos de corte donde existen mayores diferencias entre los valores.

Ambos métodos coinciden en la clasificación de las puntuaciones más bajas y más altas, pero presentan diferencias en los valores comprendidos entre 6,7 y 7,0.

Los terciles clasifican estas ciudades dentro de la categoría low, mientras que Jenks las incluye en medium. Por tanto, Jenks genera un grupo bajo más reducido y amplía el grupo medio.

A nivel cuantitativo, al comparar las categorías de ambos métodos, 333 empleados dejarían de estar en la categoría low al aplicar el método de Jenks y pasarían a la categoría medium. Esto implica que, para estos 333 empleados, dejaría de aplicarse el ajuste al alza de +0,5 asociado a la categoría low, ya que la categoría medium tiene un factor de ajuste de 0.

# Exercici 1: Exploració visual del dataset

> Aquest exercici se centra en crear una funció que ajudi a crear (i exportar) un gràfic automàticament per a cada columna del dataframe. Per exemple, un histograma/boxplot si la variable és numèrica, unes barres dels valors més freqüents si és categòrica i unes barres dels anys més freqüents si la dada està en format data.



```python
def eda(df):
    # Configuramos el estilo visual básico de Seaborn
    sns.set_theme(style="whitegrid")

    for col in df.columns:

        # excluimos estas columnas descriptivas
        if col.lower() in ['id', 'nom', 'cognoms', 'dni', 'variable']:
            continue

        # para columnas formato fecha
        if pd.api.types.is_datetime64_any_dtype(df[col]):
            anys = df[col].dt.year
            counts = anys.value_counts().sort_index()

            plt.figure(figsize=(8, 4))
            sns.barplot(x=counts.index.astype(str), y=counts.values, palette="magma", hue=counts.index.astype(str), legend=False)
            plt.title(f"Años más frecuentes: {col}")
            plt.xlabel("Año")
            plt.ylabel("Frecuencia")

        # para columnas numericas
        elif pd.api.types.is_numeric_dtype(df[col]):
            fig, axes = plt.subplots(1, 2, figsize=(12, 4))

            # Histograma
            sns.histplot(df[col].dropna(), kde=True, ax=axes[0], color="skyblue")
            axes[0].set_title(f"Histograma de {col}")

            # Boxplot
            sns.boxplot(x=df[col].dropna(), ax=axes[1], color="lightcoral")
            axes[1].set_title(f"Boxplot de {col}")

        # para columnas categoricas
        else:
            plt.figure(figsize=(8, 4))
            top_categorias = df[col].value_counts().head(10) # para mostrar los 10 valores más frecuentes
            sns.barplot(x=top_categorias.values, y=top_categorias.index.astype(str), palette="viridis", hue=top_categorias.index.astype(str), legend=False)
            plt.title(f"Categorías comunes: {col}")


        plt.tight_layout()
        plt.show()
        plt.close()
```

# Exercici 2: Optimització de rutes (problema aplicat)

> Carrega l'arxiu matriu_distancies.xlsx a pandas de manera que les files i columnes siguin les ciutats i s'eliminin "Las Palmas de Gran Canaria" i "Palma".



```python
matriu_distancies = pd.read_excel('/content/drive/MyDrive/Colab Notebooks/matriu_distancies.xlsx', index_col=0)

matriu_distancies = matriu_distancies.drop(
    columns=['Las Palmas de Gran Canaria', 'Palma'],
    index=['Las Palmas de Gran Canaria', 'Palma']
)

matriu_distancies
```

> Desenvolupa una funció que:

Rebi: una ciutat d'origen i una ciutat de destinació.

Retorni: l'ordre de les ciutats visitades (ruta) i la distància total recorreguda.

La ruta s'ha de construir seguint una estratègia senzilla: anar sempre a la ciutat no visitada més propera fins a arribar a la destinació.


```python
def calcular_ruta_cercana(origen, destino):
    ruta = [origen]
    distancia_total = 0
    ciudad_actual = origen

    # Mientras no hayamos llegado a la ciudad de destino
    while ciudad_actual != destino:
        distancias_actuales = matriu_distancies.loc[ciudad_actual].copy()

        # Eliminamos de las opciones las ciudades que YA hemos visitado en la ruta
        opciones_validas = distancias_actuales.drop(labels=ruta, errors='ignore')

        # Buscamos la ciudad no visitada que tenga la distancia más pequeña
        proxima_ciudad = opciones_validas.idxmin()
        distancia_tramo = opciones_validas.min()

        # Actualizamos la ruta, sumamos los kilómetros y cambiamos de ciudad actual
        ruta.append(proxima_ciudad)
        distancia_total += distancia_tramo
        ciudad_actual = proxima_ciudad

    return ruta, distancia_total
```

```python
ruta_final, kms_totales = calcular_ruta_cercana('Barcelona', 'Bilbao')

print(f"ORDRE RUTA: {', '.join(ruta_final)}")
print(f"DISTANCIA RECORREGUDA: {kms_totales} km")
```
