**Funcion obtenir_dades**

**Funcion validar_dades**

Comprobamos que los datos sean correctos y dentro de los rangos permitidos.

Creamos unos **patrones para el nombre y el telefono** usando el module re.

- Patron nombre: solo puede contener letras, espacios, acentos y guiones y debe haber un minimo de 2 caracteres.

- Patron telefono: longitud de 9 dígitos y debe empezar por 6 o 7.

**Validacion numerica**

Para los años de experiencia, el nivel de catalán, excel y para el título académico, se esperan digitos enteros. Trataremos de convertirlos a int, para poder usarlos en el calculo de la puntuacion.

**Validacion de rangos**

- años de experiencia: de 0 a 50

- nivel de catalán/excel/título académico: de 0 a 3


*Creamos una lista en la que se van añadiendo los errores. De esta manera, el usuario respondera a todas las preguntas y solo al final se le mostrará el cúmulo de errores si las ha habido.

````python
def validar_dades(nombre_apellidos, telefono, experiencia, catalan, excel, titulo_academico):
    import re #Se importa el modulo re para validar el formato del nombre y del telefono
    errores = [] #Contenedor para recopilar los errores mientras el candidato rellena el formulario

#Validación nombre: se especifican los caracteres autorizados y que debe haber al mínimo 2 caracteres.
    patron_nombre = r"^[A-Za-zÁÉÍÓÚáéíóúÑñ -]{2,}$"
    if not re.match(patron_nombre, nombre_apellidos):
        errores.append("El nombre solo puede contener letras, espacios, acentos y guiones (mínimo 2 caracteres).")

#Validación telefono: se determina la longitud y que debe empezar por 6 o 7)
    patron_telefono = r"[67]\d{8}"
    if not re.fullmatch(patron_telefono, telefono):
        errores.append("El teléfono debe tener exactamente 9 dígitos y empezar por 6 o 7.")

# Validación numérica
    try:
      experiencia = int(experiencia)
      catalan = int(catalan)
      excel = int(excel)
      titulo_academico = int(titulo_academico)

    except ValueError:
      errores.append("Introduzca números válidos para los años de experiencia, nivel de catalán y excel y para el título académico.")
      return errores, None, None, None, None, None, None

   # Validación de rangos
    if experiencia < 0 or experiencia > 50:
      errores.append("Debe introducir entre 0 y 50 años. Ponga 0 si lleva menos de un año.")

    if catalan < 0 or catalan > 3:
      errores.append("El nivel de catalán debe estar entre 0 y 3.")

    if excel < 0 or excel > 3:
      errores.append("El nivel de Excel debe estar entre 0 y 3.")

    if titulo_academico < 0 or titulo_academico > 3:
      errores.append("El valor del título académico debe estar entre 0 y 3.")


    return errores, nombre_apellidos, telefono, experiencia, catalan, excel, titulo_academico
````

**Funcion calcular_puntuacio**

Las variables son numeros enteros y pueden entrar en un calculo.

Guardamos la puntuacion, que luego añadiremos en la lista del candidato.

````python
def calcular_puntuacio(experiencia, catalan, excel, titulo_academico):
  puntuacion = experiencia * 1.5 + catalan + excel + titulo_academico
  return puntuacion
````

**Funcion mostrar_informe**

Debemos ordenar el informe por prioridad, y en caso de empate, por puntuación.

1 - Buscamos la puntuacion que se encuentra en el indice 6, y se ordena de mayor a menor

2 - Buscamos la prioridad que se encuentra en el indice 7, previamente se personaliza el orden de las prioridades.

3 - Para el mostrar el informe, tambien se acceden a los datos por indice.

````python
def mostrar_informe(candidatos):

    # Ordenar por puntuación (descendente)
    candidatos.sort(key=lambda c: c[6], reverse=True)

    # Ordenar por prioridad según clasificación
    orden_prioridad = ["Prioritat Alta", "Candidat adequat", "No prioritari"]
    candidatos.sort(key=lambda c: orden_prioridad.index(c[7]))

    print("\n===== INFORME FINAL DE CANDIDATOS =====\n")

    for cand in candidatos:
        print(f"Nombre: {cand[0]}")
        print(f"Telefono: {cand[1]}")
        print(f"Puntuación: {cand[6]}")
        print(f"Clasificación: {cand[7]}")
        print("---------------------------------------")
````

**Funcion main_seleccio:**

Integramos las funciones que hemos creado y establecemos el orden de ejecucion.

- Se repreguntaran los datos al candidato con un bucle While hasta que esten validos.
- Calcular puntuación de cada candidato
- Clasificar candidatos según puntuación
- Almacenar los datos del formulario en lista
- Mostrar puntuación al finalizar un registro
- Establecemos que se puedan realizar hasta 10 registros con un bucle For con range().

````python
def procesar_candidatos(lista_candidatos):

    candidatos_validos = []

    for datos in lista_candidatos:

        errores, nombre_apellidos, telefono, experiencia, catalan, excel, titulo_academico = validar_dades(*datos)

        if errores:
            print(f"\nErrores en {nombre_apellidos}:")
            for e in errores:
                print("-", e)
            continue

        puntuacion = calcular_puntuacio(
            experiencia,
            catalan,
            excel,
            titulo_academico
        )

        clasificacion = classificar_candidat(puntuacion)

        candidato = [
            nombre_apellidos,
            telefono,
            experiencia,
            catalan,
            excel,
            titulo_academico,
            puntuacion,
            clasificacion
        ]

        candidatos_validos.append(candidato)

        if len(candidatos_validos) >= 10:
            print("\nSe ha alcanzado el límite de 10 candidatos.")
            break

    return candidatos_validos
````

**función obtener_segment_i_factor()**

Creamos una lista de diccionarios "segmentos" compuesto de 1 diccionario por segmento, y que incluye 3 claves: el segmento del cliente, su mensaje asociado y su correspondiente factor.

La funcion recibe un código de comportamiento (0–4) y devuelve el diccionario correspondiente.

````python
segmentos = {
    0: {"segmento": "Navega pero no añade productos",
        "mensaje": "Aprovecha nuestras ofertas",
        "factor_codigo": 0.10},

    1: {"segmento": "Agrega productos pero no llega al checkout",
        "mensaje": "Podemos ayudarte?",
        "factor_codigo": 0.25},

    2: {"segmento": "Llega al checkout pero no paga",
        "mensaje": "Descuento valido hasta mañana",
        "factor_codigo": 0.50},

    3: {"segmento": "Abandono <24h",
        "mensaje": "Obtiene 10% cashback para tu próxima compra",
        "factor_codigo": 0.75},

    4: {"segmento": "Abandono >24h",
        "mensaje": "Aqui tienes tu cupón para tu próxima compra",
        "factor_codigo": 0.90}
}

def obtener_segment_i_factor(codi_comportament):
    return segmentos[codi_comportament]

````

**función calcular_risc()**

Esta función recibe dos parametros obtenidos de otras funciones: factor_codigo & factor_edad y los multiplica

````python
def calcular_riesgo(factor_codigo, factor_edad):
    return (factor_codigo * factor_edad)
````

**función procesar_clients()**

Esta función integra las 3 anteriores (obtenir_segment_i_factor / obtenir_factor_edat / calcular_risc) y devuelve un diccionario que incluye los nuevos datos.

Con un bucle For, para cada cliente se extraen los nuevos datos (segmento, factor_codigo, factor_edad, riesgo_final) ejecutando las 3 funciones.

Se crea para cada cliente, un diccionario que almacena sus datos, y todos los diccionarios se guardan en un diccionario global. Asi obtenemos un diccionario de diccionario.


````python
def procesar_clients(clientes):
    diccionario_final = {}

    for cliente in clientes:

        # 1) Obtener segmento, mensaje y factor del código
        info_codigo = obtener_segment_i_factor(cliente["codi_comportament"])

        # 2) Obtener factor de edad
        factor_edad = obtener_factor_edat(cliente["grup_edat"])

        #Calcular riesgo final
        riesgo = calcular_riesgo(info_codigo["factor_codigo"], factor_edad)

        #Construir diccionario
        diccionario_final[cliente["id_client"]] = {
            "id_client": cliente["id_client"],
            "nom": cliente["nom"],
            "edat": cliente["edat"],
            "grup_edat": cliente["grup_edat"],
            "codi_comportament": cliente["codi_comportament"],
            "segmento": info_codigo["segmento"],
            "mensaje": info_codigo["mensaje"],
            "factor_codigo": info_codigo["factor_codigo"],
            "factor_edad": factor_edad,
            "riesgo_final": riesgo
        }

    return diccionario_final
````

````python
def limpiar_y_normalizar(lista_bruta):

    lista_normalizada = []

    for elemento in lista_bruta:
        elemento_limpio = elemento.lower().strip().replace("|", ",")  # lower: pasa las mayúsculas a minúsculas / strip: elimina espacios / replace: cambia separadores "|" por ","

        # Dividir la cadena en elementos
        elementos = elemento_limpio.split(",")

        elementos_normalizados = []
        for elemento in elementos:
            elemento = elemento.strip()           # elimina los espacios al inicio y final de cada elemento
            elemento = elemento.replace(" ", "_") # convierte los espacios en guion bajo
            elementos_normalizados.append(elemento)

        lista_normalizada.append(elementos_normalizados)

    return lista_normalizada
````

- Devuelve una lista normalizada:

  ['maria', '45', 'tos', 'fiebre_alta', 'dolor_pecho']

 - Limpieza general:

    Pasamos todos las letras a minuscula, eliminamos los espacios y separamos los elementos por coma.

- Normalizacion de cada elemento:

  Eliminamos los espacios al inicio y final de cada elemento y convertimos el espacio entre dos palabras en un guion bajo para seguir el patron del diccionario de niveles de riesgo por sintoma.

**funcion crear_diccionario()**

* Recibe una lista normalizada como:

  ['maria', '45', 'tos', 'fiebre_alta', 'dolor_pecho']
    
* Devuelve un diccionario:

  {'nombre': 'maria','edad': '45','sintomas': ['tos', 'fiebre_alta','dolor_pecho']}

* Creamos 3 claves: nombre, edad y sintomas y se asignan los valores por indices de la lista:

    Primer indice = nombre

    Segundo indice = edad

    A partir del 3r indice (y todo lo que siga despues) = sublista de sintomas





````python
def crear_diccionario(lista_normalizada):

    nombre = lista_normalizada[0]     # el primer indice es el nombre
    edad = lista_normalizada[1]       # el segundo indice es la edad
    sintomas = lista_normalizada[2:]  # a partir del tercer indice, son los sintomas

    return {
        "nombre": nombre,
        "edad": edad,
        "sintomas": sintomas
    }
````

Sumamos los valores de los síntomas y se añade un nuevo par de clave:valor "nivel_riesgo:suma de valores" al diccionario.

````python
def calcular_nivel_riesgo(diccionario_paciente, nivells_risc):

    sintomas = diccionario_paciente["sintomas"]

    # Sumar los valores de cada síntoma
    suma_riesgo = 0
    for sintoma in sintomas:
        suma_riesgo += nivells_risc[sintoma]

    # Añadir nueva clave al diccionario
    diccionario_paciente["riesgo_total"] = suma_riesgo

    return diccionario_paciente
````

**función combinar_datos()**


Observamos que "Clients" ya es una lista de diccionarios y en el diccionario "nivells_seguro", la llave externa corresponde a la llave “codi” de "Clients".

Usaremos ese dato comun para agregar los dos conjuntos de datos. De esa forma añadiremos los 3 campos categoria, preu base y probabilitat de frau en una misma operación.

La función recibe la lista de clientes y el diccionario de niveles de seguro como parametros, y devuelve una lista de diccionario.


````python
def combinar_datos(clients, nivells_seguro):
    asegurados = []
    for client in clients:
        codi = client["codi"]  # extraemos el valor de 'codi' del diccionario del cliente
        llave = nivells_seguro[codi]  # apuntamos a la llave externa 'codi' dentro de nivells_seguro
        copia = client.copy()  # creamos una copia del diccionario del cliente
        copia.update(llave)  # se junta la copia con los datos del nivel de seguro
        asegurados.append(copia)  # se añade el diccionario enriquecido a la lista 'asegurados'
    return asegurados

asegurados = combinar_datos(clients, nivells_seguro)
````

**funcion definir_caracteres()**

Se definen los tipos de caracteres que serán incluidos en la contraseña, usaremos unas constantes definidas en el modulo string:

- ascii_lowercase: comprende las 26 letras ASCII en minusculas
- ascii_uppercase: comprende las 26 letras ASCII en mayusculas
- digits: contiene los 10 dígitos
- punctuation: contiene los caracteres de puntuación en la configuración regional

La funcion recibe los 4 tipos de caracteres posibles, y devuelve el alfabeto final (la lista de caracteres).

````python
import string
import random

def definir_caracteres(mayusculas, minusculas, numeros, signos):
  alfabeto = ""
  if mayusculas: alfabeto += string.ascii_uppercase
  if minusculas: alfabeto += string.ascii_lowercase
  if numeros: alfabeto += string.digits
  if signos: alfabeto += string.punctuation
  return alfabeto
````

**funcion asegurar_minimo()**

Usaremos la funcion random.choice de la libreria numpy, que retorna aleatoriamente uno o varios elementos.

Aseguramos que es escoja al menos un caracter de cada tipo selecionado.
   

````python
import numpy as np

def asegurar_minimo(mayusculas, minusculas, numeros, signos):
    minimos = []

    if mayusculas : minimos.append(random.choice(string.ascii_uppercase))
    if minusculas : minimos.append(random.choice(string.ascii_lowercase))
    if numeros: minimos.append(random.choice(string.digits))
    if signos: minimos.append(random.choice(string.punctuation))

    return minimos
````

**funcion generar_contrasena()**

La función recibe los 4 tipos de caracteres posibles y la longitud.

Se genera la contraseña final con:

- 1 carácter obligatorio por cada alfabeto seleccionado

- el resto de caracteres aleatorios

Y se mezcla todo para evitar patrones.


````python
def generar_contrasena(longitud=12, mayusculas=True, minusculas=True, numeros=True, signos=False):
  alfabeto = definir_caracteres(mayusculas, minusculas, numeros, signos)
  minimos = asegurar_minimo(mayusculas, minusculas, numeros, signos)

  faltantes = longitud - len(minimos)
  contrasena = minimos

  for _ in range(faltantes):
    contrasena.append(random.choice(alfabeto))
    random.shuffle(contrasena)

  return "".join(contrasena)
````

**funcion parsear_linea()**

Tenemos actualmente 3 columnas de las que queremos extraer el nombre de los equipos y los goles de cada uno. Accederemos a cada elemento de las filas por su indice.

Extraemos y limpiamos los nombres de los equipos con strip(), asi evitamos que surjan duplicados.
Limpiamos el resultado y se "desempaqueta": La cadena se separa donde se encuentra el guion y se devuelve una lista de 2 elementos.
Ex: " 0-1 " se transforma en ['2', '1'].

Finalmente se convierten estos elementos a numero entero.

Se devuelve una tupla con los 4 valores: equipoA, equipoB, golesA, golesB.

````python
def parsear_linea(fila):

    equipoA = fila[0].strip()
    resultado = fila[1].strip()
    equipoB = fila[2].strip()

    golesA, golesB = resultado.split('-')

    golesA = int(golesA)
    golesB = int(golesB)

    return equipoA, equipoB, golesA, golesB
````

**funcion def actualizar_estadisticas()**

Comprobamos si los equipos existen en el diccionario. Si es nuevo, se añade dejando los valores (puntos, goles, partidos) a 0

Sumamos los goles generados por cada equipo en "goles_favor" y, a su vez, los añadimos a la cuenta de "goles_contra" del equipo rival.

Mediante condiciones, realizamos la suma de puntos (+3,+1,+0) segun cada escenario, y actualizamos los resultados (partido ganado, perdido o empatado)

````python
def actualizar_estadisticas(stats, equipoA, equipoB, golesA, golesB):
    # Iteramos sobre los 2 equipos de cada linea
    for equipo in (equipoA, equipoB):
        # Si el equipo no existe en el diccionario, se crea con todos los valores a 0.
        if equipo not in stats:
            # Creacion de las llaves del diccionario
            stats[equipo] = {
                "goles_favor": 0,
                "goles_contra": 0,
                "ganados": 0,
                "empatados": 0,
                "perdidos": 0,
                "puntos": 0
            }

    # Suma de los goles
    stats[equipoA]["goles_favor"] += golesA
    stats[equipoA]["goles_contra"] += golesB
    stats[equipoB]["goles_favor"] += golesB
    stats[equipoB]["goles_contra"] += golesA

    # Calculo de puntos (+3,+1,+0) y resultados: partido ganado, perdido o empatado

    # Escenario equipoA ha ganado
    if golesA > golesB:
        stats[equipoA]["ganados"] += 1
        stats[equipoB]["perdidos"] += 1
        stats[equipoA]["puntos"] += 3
    # Escenario equipoB ha ganado
    elif golesB > golesA:
        stats[equipoB]["ganados"] += 1
        stats[equipoA]["perdidos"] += 1
        stats[equipoB]["puntos"] += 3
    # Escenario en caso de empate
    else:
        stats[equipoA]["empatados"] += 1
        stats[equipoB]["empatados"] += 1
        stats[equipoA]["puntos"] += 1
        stats[equipoB]["puntos"] += 1

    return stats
````

````python
def crear_dataframe(stats):
    df = pd.DataFrame.from_dict(stats, orient="index") # convertimos el diccionario en dataframe
    df.index.name = "equipo" # añadimos el nombre de columna "equipo"
    df= df.sort_values(by="puntos", ascending=False) # ordenamos las filas por la columna "puntos" de mayor a menor
    return df
````

**funcion mostrar_resultados()**

Ya tenemos la tabla ordenada por puntuación.

Recorremos el indice empezando por la posicion 1. La funcion enumarate() nos devuelve el indice con su correspondiente valor.


````python
def mostrar_resultados(df):
    mas_goleador = df.loc[df["goles_favor"].idxmax()] # Devuelve el índice de la fila con más goles a favor
    mas_goleado = df.loc[df["goles_contra"].idxmax()] # Devuelve el índice de la fila con más goles en contra

    print("\n CLASIFICACION")
    display(df)

    print("\n Equipo más goleador:", mas_goleador.name, "-", mas_goleador["goles_favor"], "goles")
    print(" Equipo más goleado:", mas_goleado.name, "-", mas_goleado["goles_contra"], "goles")
    print(" \n Clasificación final:\n")

     # Clasificación final en texto: recorre el índice ya ordenado
    for posicion, equipo in enumerate(df.index, start=1):
        print(f"{posicion}. {equipo}")
````
