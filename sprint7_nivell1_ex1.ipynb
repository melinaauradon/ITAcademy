**Funcion obtenir_dades**

Se crea el formulario de inscripcion solicitando los datos vía input y guardamos las entradas para que sean nuestras variables.


````python
def obtenir_dades():

  nombre_apellidos = input("Introducza su nombre y apellidos: ")
  telefono = input("Introducza los 9 dígitos de su número de teléfono: ")
  experiencia = input("Cuántos años lleva trabajando en la empresa? ")
  catalan = input("Introducza el dígito que corresponda a sus conocimientos de Catalán\n\t0 = sense coneixement\n\t1 = fins a nivell B1\n\t2 = nivell B2\n\t3 = C1 o superior\n")
  excel = input("Introducza el dígito que corresponda a su dominio de Excel\n\t0 = nivell bàsic\n\t1 = ús de funcions\n\t2 = taules dinàmiques\n\t3 = Visual Basic i Macros\n")
  titulo_academico = input("Introducza el dígito que corresponda con su título académico\n\t0 = ESO\n\t1 = FP\n\t2 = Universitat\n\t3 = Màster o superior\n")

  return nombre_apellidos, telefono, experiencia, catalan, excel, titulo_academico
````

**Funcion validar_dades**

Comprobamos que los datos sean correctos y dentro de los rangos permitidos.

Creamos unos **patrones para el nombre y el telefono** usando el module re.

- Patron nombre: solo puede contener letras, espacios, acentos y guiones y debe haber un minimo de 2 caracteres.

- Patron telefono: longitud de 9 dígitos y debe empezar por 6 o 7.

**Validacion numerica**

Para los años de experiencia, el nivel de catalán, excel y para el título académico, se esperan digitos enteros. Trataremos de convertirlos a int, para poder usarlos en el calculo de la puntuacion.

**Validacion de rangos**

- años de experiencia: de 0 a 50

- nivel de catalán/excel/título académico: de 0 a 3


*Creamos una lista en la que se van añadiendo los errores. De esta manera, el usuario respondera a todas las preguntas y solo al final se le mostrará el cúmulo de errores si las ha habido.

````python
def validar_dades(nombre_apellidos, telefono, experiencia, catalan, excel, titulo_academico):
    import re #Se importa el modulo re para validar el formato del nombre y del telefono
    errores = [] #Contenedor para recopilar los errores mientras el candidato rellena el formulario

#Validación nombre: se especifican los caracteres autorizados y que debe haber al mínimo 2 caracteres.
    patron_nombre = r"^[A-Za-zÁÉÍÓÚáéíóúÑñ -]{2,}$"
    if not re.match(patron_nombre, nombre_apellidos):
        errores.append("El nombre solo puede contener letras, espacios, acentos y guiones (mínimo 2 caracteres).")

#Validación telefono: se determina la longitud y que debe empezar por 6 o 7)
    patron_telefono = r"[67]\d{8}"
    if not re.fullmatch(patron_telefono, telefono):
        errores.append("El teléfono debe tener exactamente 9 dígitos y empezar por 6 o 7.")

# Validación numérica
    try:
      experiencia = int(experiencia)
      catalan = int(catalan)
      excel = int(excel)
      titulo_academico = int(titulo_academico)

    except ValueError:
      errores.append("Introduzca números válidos para los años de experiencia, nivel de catalán y excel y para el título académico.")
      return errores, None, None, None, None, None, None

   # Validación de rangos
    if experiencia < 0 or experiencia > 50:
      errores.append("Debe introducir entre 0 y 50 años. Ponga 0 si lleva menos de un año.")

    if catalan < 0 or catalan > 3:
      errores.append("El nivel de catalán debe estar entre 0 y 3.")

    if excel < 0 or excel > 3:
      errores.append("El nivel de Excel debe estar entre 0 y 3.")

    if titulo_academico < 0 or titulo_academico > 3:
      errores.append("El valor del título académico debe estar entre 0 y 3.")


    return errores, nombre_apellidos, telefono, experiencia, catalan, excel, titulo_academico
````

**Funcion calcular_puntuacio**

Las variables son numeros enteros y pueden entrar en un calculo.

Guardamos la puntuacion, que luego añadiremos en la lista del candidato.

````python
def calcular_puntuacio(experiencia, catalan, excel, titulo_academico):
  puntuacion = experiencia * 1.5 + catalan + excel + titulo_academico
  return puntuacion
````

**Funcion classificar_candidat**

Si ninguna de las dos primeras condiciones "if/elif" existen , se aplica "else".

Guardamos la clasificacion, que luego añadiremos a la lista del candidato.





````python
def classificar_candidat(puntuacion):

    if puntuacion >= 12:
        return "Prioritat Alta"

    elif puntuacion >= 8:
        return "Candidat adequat"

    else:
        return "No prioritari"

    return clasificacion
````

**Funcion mostrar_informe**

Debemos ordenar el informe por prioridad, y en caso de empate, por puntuación.

1 - Buscamos la puntuacion que se encuentra en el indice 6, y se ordena de mayor a menor

2 - Buscamos la prioridad que se encuentra en el indice 7, previamente se personaliza el orden de las prioridades.

3 - Para el mostrar el informe, tambien se acceden a los datos por indice.

````python
def mostrar_informe(candidatos):

    # Ordenar por puntuación (descendente)
    candidatos.sort(key=lambda c: c[6], reverse=True)

    # Ordenar por prioridad según clasificación
    orden_prioridad = ["Prioritat Alta", "Candidat adequat", "No prioritari"]
    candidatos.sort(key=lambda c: orden_prioridad.index(c[7]))

    print("\n===== INFORME FINAL DE CANDIDATOS =====\n")

    for cand in candidatos:
        print(f"Nombre: {cand[0]}")
        print(f"Telefono: {cand[1]}")
        print(f"Puntuación: {cand[6]}")
        print(f"Clasificación: {cand[7]}")
        print("---------------------------------------")
````

**Funcion main_seleccio:**

Integramos las funciones que hemos creado y establecemos el orden de ejecucion.

- Establecemos que se puedan realizar hasta 10 registros con un bucle For con range().
- Se repreguntaran los datos al candidato con un bucle While hasta que esten validos.
- Calcular puntuación de cada candidato
- Clasificar candidatos según puntuación
- Almacenar los datos del formulario en lista
- Mostrar puntuación al finalizar un registro
- Generar informe final

````python
def main_seleccio():
#Los datos se almacenan en listas
    candidatos = []
#Establecemos que se puedan realizar hasta 10 registros
    for i in range(10):
        while True: #no pasamos al siguiente candidato hasta que los datos sean válidos
            datos = obtenir_dades() #1 se piden los datos
            errores, nombre_apellidos, telefono, experiencia, catalan, excel, titulo_academico = validar_dades(*datos) #2 se validan los datos

            if errores:
                for e in errores:
                    print("-", e)
                print("Por favor, introduzca los datos nuevamente.")
                continue   #3 Si hay errores, se muestran los mensajes de error y se repite el bucle #1#2#3

            break  #Si no hay errores, se para el bucle

        puntuacion = calcular_puntuacio(experiencia, catalan, excel, titulo_academico) #llamamos la funcion que calcula la puntuacion
        clasificacion = classificar_candidat(puntuacion) #llamamos la funcion que establece la clasificacion

        #estructura de la lista
        candidato = [
            nombre_apellidos,
            telefono,
            experiencia,
            catalan,
            excel,
            titulo_academico,
            puntuacion,
            clasificacion
        ]

        candidatos.append(candidato) #se guarda el candidato
        mostrar_individual(nombre_apellidos, puntuacion) #llamamos la funcion que confirma el registro y muestra la puntuacion

    mostrar_informe(candidatos) #llamamos la funcion que devuelve el informe para RRHH

    return candidatos
````
